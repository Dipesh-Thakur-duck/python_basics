## Python Function Arguments
The terms parameter and argument can be used for the same thing: information that are passed into a function.  
From a function's perspective:  
- A parameter is the variable listed inside the parantheses in the function definition.
- An argument is the actual value that is sent to the function when it is called.

````python
def my_name(name): # name is a parameter
    print('Hello', name)

my_name('Dipesh') # Dipesh is an argument
````

By default, a function must be called with the correct number of arguments.

### Default Parameter Values

````python
def my_name(name='Dipesh'):
    print('Hello', name)

my_name()
my_name('Krishna')
````

### Keyword Arguments

````python
def my_self(name, age):
    print('My name is', name)
    print('I am', age)
my_self(name='Dipesh', age='22')
````

The order of arguments does not matter with keyword arguments.  
The phrase keyword arguments is often shortened to kwargs in Python documentation.

### Positional Arguments
When we call a function with arguments without using keywords, they are called positional arguments.  
The order matters with positional arguments.

### Mixing Positional and Keyword Arguments

````python
def my_self(name, age, sex):
    print('My name is '+ name + '. I am ' + age + '. I am '+ sex);
my_self('Dipesh', '22', 'Male')
````

Positional arguments work with the principle 'fill in the blanks in the order'. Once you use a keyword argument before positional argument, python can't then figure out where an unlabeled value should go.

### Positional-Only Arguments

````python
def my_name(name, /): # Anything before / needs to be positional
    print('Hello', name)
my_name('Dipesh')
````

### Keyword-Only Arguments

````python
def my_name(*, name): # anything after * is keyword arguments
    print('Hello', name)
my_name(name='Dipesh')
````

### Combining Positional-Only and Keyword-Only

````python
def my_function(a, b, /, *, c, d):
  return a + b + c + d

result = my_function(5, 10, c = 15, d = 20)
print(result)
````

### Arbitrary Arguments - *args

If you do not know how many arguments will be passed into your function, add a * before the parameter name.  
This way, the function will receive a tuple of arguments and can access the items accordingly.

````python
def my_friends(*friends):
    print('The youngest friend is ' + friends[1])
my_friends("Prashant", "Sima", "Kusum")
````

### Arbitrary Keyword Arguments - **kwargs

If you do not know how many keyword arguments will be passed into your function, add two asterisks ** before the parameter name.

````python
def my_friend(**friends):
    print('Hello ' + friends["BestEver"])
my_friend(BestEver= 'Prashant', RegularClose= 'Sima')
````

### Combining *args and **kwargs
You can use both *args and **kwargs in the same function.  
The order must be:
- regular parameters
- *args
- **kwargs

````python
def my_function(title, *args, **kwargs):
  print("Title:", title)
  print("Positional arguments:", args)
  print("Keyword arguments:", kwargs)

my_function("User Info", "Emil", "Tobias", age = 25, city = "Oslo")
````

### Unpacking Arguments

````python
def my_function(a, b, c):
  return a + b + c

numbers = [1, 2, 3]
result = my_function(*numbers) # Same as: my_function(1, 2, 3)
print(result)

def my_function(fname, lname):
  print("Hello", fname, lname)

person = {"fname": "Emil", "lname": "Refsnes"}
my_function(**person) # Same as: my_function(fname="Emil", lname="Refsnes")
````

## Python Scope

````python
def myfunc():
  x = 300
  print(x)

myfunc()

# A variable created inside a function is only available inside that function. It has local scope.
````

````python
x = 300

def print_x():
    print(x)
print_x()

# x here has global scope, available everywhere.
````

### Global Keyword

````python
x = 300
def myfunc():
    global x # To change the value of a global variable inside a function, use global keyword
    x = 200

myfunc()
print(x)
````

### Nonlocal Keyword

````python
def myfunc1():
  x = "Jane"
  def myfunc2():
    nonlocal x
    x = "hello"
  myfunc2()
  return x

print(myfunc1())

## The nonlocal keyword makes the variable belong to the outer function
````

### The LEGB Rule
Python follows the LEGB rule when looking up variable names, and searches for them in this order:  
- Local: Inside the current function
- Enclosing: Inside enclosing functions (from inner to outer)
- Global: At the top level of the module
- Built-in: In python's built in namespace

## Python Decorators

### The core idea: functions are just values
In python, a function is an object, like a string or a list. That means you can:

````python
def greet():
    return "hello"

x = greet  # no parentheses! we're not calling it, just storing it
print(x())  # "hello"
````

### Second idea: functions can be defined inside other functions

````python
def outer():
    def inner():
        return "hi from inner"
    return inner  # returning the function itself, not calling it

f = outer()   # f is now the "inner" function
print(f())    # "hi from inner"
````

### Third idea: a function can take a function as input, and return a modified version

````python
def my_decorator(func):
    def wrapper():
        print("Before the function runs")
        func()  # call the original function
        print("After the function runs")
    return wrapper

def say_hello():
    print("Hello!")

# Manually "decorating":
say_hello = my_decorator(say_hello)

say_hello()
````

### Final Syntax

````python
def changecase(func):
    def myinner():
        return func().upper()
    return myinner

@changecase
def myfunction():
    return "Hello Sally"

print(myfunction())
````

Bascially:
```python
myfunction = changecase(myfunction)
```

### Arguments in the Decorated Function

````python
def changecase(func):
  def myinner(x):
    return func(x).upper()
  return myinner

@changecase # decorator function
def myfunction(nam): # decorated function
  return "Hello " + nam

print(myfunction("John"))
````

Functions that require arguments can also be decorated, just make sure you pass the arguments to the wrapper function.

### *args and **kwargs in decorator function
Sometimes the decorator function has no control over the arguments passed from decorated function. To solve this used **args and **kwargs.

````python
def changecase(func):
  def myinner(*args, **kwargs):
    return func(*args, **kwargs).upper()
  return myinner

@changecase
def myfunction(nam):
  return "Hello " + nam

print(myfunction("John"))
````

### Decorator with Arguments

````python
def changecase(n):
  def changecase(func):
    def myinner():
      if n == 1:
        a = func().lower()
      else:
        a = func().upper()
      return a
    return myinner
  return changecase

@changecase(1)
def myfunction():
  return "Hello Linus"

print(myfunction())
````

#### Frist, notice the naming trap
```python
def changecase(n):
    def changecase(func):
        ...
    return changecase
```
The inner function is also called changecase. This isn't an error - Python allows it because the inner one lives in its own local scope. But it's confusing to read, so mentally rename it:  
```python
def changecase(n):
  def actual_decorator(func):
    def myinner():
      if n == 1:
        a = func().lower()
      else:
        a = func().upper()
      return a
    return myinner
  return actual_decorator
```
#### What @changecase(1) really means
Normally @decorator means "pass the function into decorator'. But here you're not writing @changecase - you're writing @changecase(1), which is a function call that happens first, and its result becomes the decorator. So the manual (non-@) version is:  
```python
def myfunction():
    return "Hello Linus"

myfunction = changecase(1)(myfunction)

print(myfunction())
```
Two separate calls are stacked here: changecase(1) then (myfunction). Let's trace each.  
#### Step 1: changecase(1) runs
```python
def changecase(n):      # n = 1
  def actual_decorator(func):
      ...
  return actual_decorator
```
- n is set to 1 inside this call.
- actual_decorator is defined - and because of closures, it remembers n=1 even after changecase finishes.
- changecase(1) returns actual_decorator (not yet called).
#### Step 2: actual_decorator(myfunction) runs
Continuing the line changecase(1)(myfunction), we now call the returned actual_decorator with myfunction as func:  
```python
def actual_decorator(func):   # func = myfunction
    def myinner():
        if n == 1:              # n is still 1, remembered from step 1
            a = func().lower()
        else:
            a = func().upper()
        return a
    return myinner
```
- func is set to the original myfunction.
- myinner is defined : it remembers both func (=myfunction and n = 1)
- actual_decorator returns myinner (not called yet)
So now: actual_decorator(myfunction)->myinner (a function, remembering func=myfunction and n=1_
#### Step 3: reassignment
myfunction = changecase(1)(myfunction)
myfunction = myinner
#### Step 4: print(myfunction())
This actually calls myinner().

### Preserving Function Metadata
Functions in Python has metadata that can be accessed using the `__name__` and `__doc__` attributes.

````python
def myFunction():
    return "Have a great day"
print(myFunction.__name__)
````

But, when a function is decorated, the metadata of the original function is lost.

````python
def changecase(func):
    def inner():
        return func().upper()
    return inner

@changecase
def myname():
    return 'Dipesh Thakur'
print(myname())
print(myname.__name__)
````

To fix this, Python has a built-in function called functools.wraps that can be used to preserve the original function's name and docstring.

````python
import functools

def changecase(func):
    @functools.wraps(func)
    def inner():
        return func().upper()
    return inner

@changecase
def myname():
    return 'Dipesh Thakur'
print(myname())
print(myname.__name__)
````

## Python Lambda

### Lambda Functions
A lambda function is a small anonymous function.  
A lambda function can take any number of arguments, but can only have one expression.

````python
x = lambda a: a + 5
print(x(5))

y = lambda a,b : a if a > b else b
print(y(5, 10))
````

### Why use lambda functions?
It shines when you need a quick function inline.

#### Using Lambda with map()

````python
numbers = [1, 2, 3, 4]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
````

#### Using Lambda with filter()

````python
numbers = [2, 3, 4, 5, 6]
even_numbers = list(filter(lambda num: num % 2 == 0, numbers))
print(even_numbers)
````

#### Using Lambda with sorted()

````python
students = [('Dipesh', 22), ('Sarita', 23), ('Sima', 19)]
sorted_students = sorted(students, key = lambda x: x[1])
for x in sorted_students:
    print(x[0])
````

### Passing a lambda function

````python
def myfunc(n):
    return lambda a: a * n
double = myfunc(2)
print(double(11))
````

double get the lambda function as value which using closure remembers value of n and when we pass 11, it's doubled.

## Python Recursion

### Recursion
It is when a function calls itself.  
<span style="color:red">Warning!:</span> The developer should be very careful with recurssion as it can be quite easy to slip into writing a function which never terminates.

````python
def countdown(n):
    if n <= 0:
        print('Done!')
    else:
        print(n)
        countdown(n - 1)
countdown(5)
````

Every recursive function must have two parts:
- A base case: A condition that stops the recursion
- A recursive case: The function calling itself with a modified argument
Without a base case, the function would call itself forever, causing a stack overflow error.

#### Recursion Depth Limit
Python has a limit on how deep recursion can go. The default limit is usually around 1000 recursive calls.

## Python Generators
Generators are functions that can pause and resume their execution.  
Generators allow you to iterate over data without storing the entire dataset in memory.  

````python
def my_generator():
    yield 1
    yield 2
    yield 3
for value in my_generator():
    print(value)
````

### The yield keyword
The yield keyword is what makes a function a generator.  
When yield is encountered, the function's state is saved, and the value is returned. The next time the generator is called, it continues from where it left off.

### Generators saves memory
Generators are memory-efficient because they generate values on-the-fly instead of storing everything in memory.

### Using next() with generators

````python
def simple_gen():
    yield "Emil"
    yield "Tobias"
    yield "Linus"
gen = simple_gen()
print(next(gen))
print(next(gen))
print(next(gen))
# print(next()) This will raise StopIteration exception
````

You can't do this:  
```python
print(next(simple_gen())
```
Because each time you pass the function to next, a new generator object is created and hence only the first yield is returned. Whereas when the generator is saved in a variable, it's resused.

### Generator Expression

````python
# List comprehension - creates a list
list_comp = [x * x for x in range(5)]
print(list_comp)

# Generator expression - creates a generator
gen_exp = (x * x for x in range(5))
print(gen_exp)
print(list(gen_exp))
````

### Generator Methods
#### send() Method
The send() method allows you to send a value to the generator. 
```python
def echo_generator():
  while True:
    received = yield
    print("Received:", received)

gen = echo_generator()
next(gen)  # Prime the generator
gen.send("Hello")
gen.send("World")
```
##### Step 1: gen = echo_generator()
This just creates the generator object.  
##### Step 2: next(gen) - this is the priming
Calling `next(gen)` starts running the function. It enters the while True: loop, hits `received = yield`, and pauses right there, waiting to receive a value. At this point, `received` hasn't received been assigned anything yet - execution is frozen exactly at the `yield` keyword.  
This step is called priming because it advances the generator from "not started" to "paused at the first yield, ready to accept input". Without doing this, the generator hasn't reached a `yeild` statement yet, so there's nowhere for a value to be "sent into".  
##### Step 3: gen.send('Hello')
- Delivers `"Hello"` into the paused `yield`expression, so `received = "Hello"`
- Resumes execution from there
So now:
```python
received = "Hello"
print("Received:", received)   # prints "Received: Hello"
```
Then the loop goes back to `while True:`, hits `yield` again, and pauses once more - waiting for the next `send()`.

### close() Method
The `close()` method stops the generator.

````python
def my_gen():
  try:
    yield 1
    yield 2
    yield 3
  finally:
    print("Generator closed")

gen = my_gen()
print(next(gen))
gen.close()
````
