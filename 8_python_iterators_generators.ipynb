# Python Iterators and Generators

## Iterable vs Iterator

```python
nums = [1, 2, 3]
# next(nums)  # would raise TypeError: 'list' object is not an iterator

it = iter(nums)      # get the iterator from the iterable
print(next(it))       # 1
print(next(it))       # 2
print(next(it))       # 3
# print(next(it))     # would raise StopIteration
```

iter() creates an iterator from iterable.  
- nums(the list) = a book. You can open it and read it, but the book itself doesn't "remember" what page you're on.
- it = iter(nums) = a bookmark. It's a separate object that tracks your current position in the book.
- next(it) = "turn to the next page and show me what's there"

## What a for loop does under the hood

```python
nums = [1, 2, 3]
it = iter(nums)
while True:
    try:
        x = next(it)
    except StopIteration:
        break
    print(x)
```

## Generator function

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i
        i += 1

gen = count_up_to(3)
print(gen)          # <generator object count_up_to at 0x...>
print(next(gen))     # 1
print(next(gen))     # 2
print(next(gen))     # 3
# print(next(gen))   # would raise StopIteration
```

## Generator expression

```python
gen = (x**2 for x in range(10))
print(gen)

for val in gen:
    print(val)
```

## Infinite Generator

```python
def infinite_counter():
    i = 0
    while True:
        yield i
        i += 1

counter = infinite_counter()
for _ in range(5):
    print(next(counter))
```

## One shot exhaustion

```python
gen = (x**2 for x in range(3))
print(list(gen))    # [0, 1, 4]
print(list(gen))    # [] -- already exhausted
```

list() forces the generator to run to completion - it keeps calling next(gen) internally until StopIteration is raised, collecting each value along the way.  

Afterwards, the generator object is now sitting at the end. Its internal position has moved past the last value.  There's nothing left for it to produce, and it has no way to "reset" back to the start.  

In third line, list() calls gen object again with next(gen) which immediately raises StopIteration because there's nothing left. list() catches the exception and gives an empty list.
