# Python Try Except
The <mark>try</mark> block lets you text a block of code for errors.  
The <mark>except</mark> block lets you handle the error.  
The <mark>else</mark> block lets you execute code when there is no error.  
The <mark>finally</mark> block lets you execute code, regardless of the result of the try and except blocks.

```python
try:
    print(x)
except:
    print("An exception occured")
```

```python
try:
    print(x)
except NameError:
    print('Variable x is not defiend')
except:
    print('Something else went wrong')
```

## Else
You can use the else keyword to define a block of code to be executed if no errors were raised.

```python
try:
    print('Hello')
except:
    print('Something went wrong')
else:
    print('Nothing went wrong')
```

## Finally
The finally block, if specified, will be executed regardless if the try block raises an error or not.

```python
try:
    print(x)
except:
    print('Something went wrong')
finally:
    print('The try...except is finished')
```

## Raise an exception
To throw (or raise) an exception, use the raise keyword.

```python
x = -1

if x < 0:
    raise Exception('Sorry, no numbers below zero')
```
