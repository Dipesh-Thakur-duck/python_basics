# Python OOP

## OOP
### What is OOP?
OOP stands for Object-Oriented Programming.  
Python supports object-oriented programming, allowing you to structure your code using classes and objects for better organization and reusability.
### Advantages of OOP
- Provides a clear structure to programs
- Makes code easier to maintain, reuse, and debug
- Helps keep your code DRY (Don't Repeat Yourself)
- Allows you to build reusable applications with less code
### What are Classes and Objects?
Class is a blueprint/template for an object that defines what it is (its structure/data) and what it can do (its valid operations).  
Ojbect is an instance of a class with its own identity (a memory address), its own state (values in its attributes) but shares the behavior (methods) defined by its class.

## Python `__init__()` Method

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def greet(self):
        return f'Hello, I am {self.name}. I am {self.age}.'

p1 = Person('Dipesh' , 22)
print(p1.greet())
```

The `__init__()` method is used to assign values to object properties, or perfrom operations that are necessary when the object is being created.

## Python self Parameter
The <mark>self</mark> parameter is a reference to the current instance of the class.  
It is used to access properties and methods that belong to the class.  
Without <mark>self</mark>, Python would not know which object's properties you want to access.

## The `__str__()` Method
The <mark>`__str__()`</mark> method is a special method that controls what is returned when the object is printed.

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __str__(self):
        return f"I am {self.name}. I am {self.age}."
p1 = Person('Emil', 36)
print(p1.__str__())
```

## Python Inheritance
Inheritance allows us to define a class (child class) that inherits all the methods and properties from another class (parent class).

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self, name, age):
        Person.__init__(self, name, age)

x = Student('Dipesh', 22)
print(f'I am {x.name}. I am {x.age}.')
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
class Student(Person):
    def __init__(self, name, age):
        super().__init__(name, age)

x = Student('Dipesh', 22)
print(f'I am {x.name}. I am {x.age}.')
```

When self is passed to the constructor of parent class in child class, python knows which object is using its parent class properties. `super()` method implicitly passes the self.

## Python Polymorphism
When same method name has different behavior depending on the actuall class of the object, its called polymorphism.

```python
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Drive!")

class Boat:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Sail!")

class Plane:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model

  def move(self):
    print("Fly!")

car1 = Car("Ford", "Mustang")       #Create a Car object
boat1 = Boat("Ibiza", "Touring 20") #Create a Boat object
plane1 = Plane("Boeing", "747")     #Create a Plane object

for x in (car1, boat1, plane1):
  x.move()
```

## Python Encapsulation
Encapsulation is when you bundle data and methods in a manner where restrictions are imposed on when data change is premissible. This gives you a single, controlled pathway for how an objects' data is allowed to change.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit must be positive")
        self.__balance += amount

    def withdraw(self, amount):
        if amount > self.__balance:
            raise ValueError("Insufficient funds")
        self.__balance -= amount

    def get_balance(self):
        return self.__balance

acc = BankAccount("Dipesh", 1000)
acc.deposit(500)
print(acc.get_balance())   # 1500

acc.withdraw(10000)
# ValueError: Insufficient funds
```

### Traditional getter/setter

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks          # private (name-mangled)

    def get_marks(self):              # getter
        return self.__marks

    def set_marks(self, value):       # setter
        if value < 0 or value > 100:
            raise ValueError("Marks must be between 0 and 100")
        self.__marks = value

s = Student("Dipesh", 85)

print(s.get_marks())      # 85
s.set_marks(92)
print(s.get_marks())      # 92

s.set_marks(150)
# ValueError: Marks must be between 0 and 100
```

### @property version (same protection, cleaner syntax)

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = marks

    @property
    def marks(self):                  # getter
        return self.__marks

    @marks.setter
    def marks(self, value):           # setter
        if value < 0 or value > 100:
            raise ValueError("Marks must be between 0 and 100")
        self.__marks = value

s = Student("Dipesh", 85)

print(s.marks)        # 85    — looks like plain attribute access
s.marks = 92           # looks like plain assignment...
print(s.marks)         # 92    # ...but setter validation runs underneath

s.marks = 150
# ValueError: Marks must be between 0 and 100
```

### Name Mangling

It is how Python implements private properties and methods.  
When you use double underscores `__`, Python automatically renames it internally by adding `_ClassName` in front.  
For example, `__age` becomes `_Person__age`.

## Python Inner Classes

An inner class is a class defined inside another class. The inner class can access the properties and methods of the outer class.  
Inner classes are useful for grouping classes that are only used in one place, making your code more organized.

```python
class Outer:
    def __init__(self):
        self.name = "Outer"
    class Inner:
        def __init__(self):
            self.name = "Inner"
        def display(self):
            print('Hello from inner class')
outer = Outer()
inner = outer.Inner()
inner.display()
```

### Accessing outer class from inner class

```python
class Outer:
  def __init__(self):
    self.name = "Emil"

  class Inner:
    def __init__(self, outer):
      self.outer = outer

    def display(self):
      print(f"Outer class name: {self.outer.name}")

outer = Outer()
inner = outer.Inner(outer)
inner.display()
```

```python
# Practical Example
class Car:
  def __init__(self, brand, model):
    self.brand = brand
    self.model = model
    self.engine = self.Engine()

  class Engine:
    def __init__(self):
      self.status = "Off"

    def start(self):
      self.status = "Running"
      print("Engine started")

    def stop(self):
      self.status = "Off"
      print("Engine stopped")

  def drive(self):
    if self.engine.status == "Running":
      print(f"Driving the {self.brand} {self.model}")
    else:
      print("Start the engine first!")

car = Car("Toyota", "Corolla")
car.drive()
car.engine.start()
car.drive()
```

```python
# Multiple Inner class
class Computer:
  def __init__(self):
    self.cpu = self.CPU()
    self.ram = self.RAM()

  class CPU:
    def process(self):
      print("Processing data...")

  class RAM:
    def store(self):
      print("Storing data...")

computer = Computer()
computer.cpu.process()
computer.ram.store()
```
