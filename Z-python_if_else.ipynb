# Python If Statement

```python
a = 200
b = 33
if b > a:
    print('b is greater than a')
elif a == b:
    print('a and b are equal')
else:
    print('a is greater than b')
```

## Short Hand If

```python
a = 5
b = 2
if a > b: print('a is greater than b')
```

## Short Hand If...Else

```python
a = 2
b = 330
print('A') if a > b else print('B') # conditional expression (sometimes known as a "ternary operator")
```

## Assign a Value With If...Else

```python
a = 10
b = 20
bigger = a if a > b else b
print('Bigger is', bigger)
```

## Multiple Conditions on One Line

```python
a = 330
b = 330
print('A') if a > b else print('=') if a == b else print('B')
```

## When to Use Shorthand If
- The conditions and actions are simple
- It improves code readability
- You want to make a quick assignment based on a condition

## Python Pass 

```python
a = 33
b = 220
if b > a:
    pass
```

if statements cannot be empty, but if you for some reason have an if statement with no content, put in the pass statement to avoid gettting an error.
