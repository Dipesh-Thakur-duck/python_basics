# Python range
The built-in range() function returns an immutable sequence of numbers, commonly used for looping a specific number of times.  
Syntax:
```python
range(start, stop, step)
```

````python
for i in range(3): # start defaults to 0 and the only argument represents the stop value
    print(i, end=" ")
print()
for i in range(2, 5): # start and stop value
    print(i, end=" ")
print()
for i in range(1, 9, 3): # start, stop and step value
    print(i, end=" ")
    
````

## Using List to Display Ranges
The range object is a data type that represents an immutable sequence of numbers, and it is not directly displayable. Therefore ranges are often converted to lists for display.

````python
print(list(range(5)))
````

## Slicing Ranges
Like other sequences, ranges can be sliced to extract a subsequence.

````python
r = range(10)
print(r[2])
print(type(r[2])) # sequence itself is of type range, but the elements it produces are plain int
print(r[:3]) # returns a new range object
````

## Membership Testing
Ranges support membership testing with the in operator.

````python
r = range(0, 10, 2)
print(6 in r)
print(7 in r)
````

## Length
Range supports the len() function to get the number of elements in the range.

````python
r = range(0, 10)
print(len(r))
````
